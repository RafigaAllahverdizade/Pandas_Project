### [C2] What are the top 20 items that are reordered most frequently (display their names and product IDs)?

```python
df_reordered_products = df[df["reordered"] == 1]
df_reordered_products = df_reordered_products.groupby("product_name")["product_id"].count()
df_reordered_products = df_reordered_products.reset_index()
df_reordered_products.rename(columns = {"product_id": "product_count"}, inplace = True)
df_reordered_products.head(20)
```

### [C3] For each product, what proportion of its orders are reorders?

```python
df_reorders = df.groupby("product_id")["reordered"].mean()
df_reorders = df_reorders.reset_index()
df_reorders.rename(columns = {"reordered": "reorder_rate"}, inplace = True)
df_reorders
```

```python
df.head()
```

### [C4] For each customer, what proportion of their products ordered are reorders?

```python
df_customer_reorders = df.groupby("user_id")["reordered"].mean()
df_customer_reorders = df_customer_reorders.reset_index()
df_customer_reorders.rename(columns = {"reordered": "reorder_rate"}, inplace = True)
df_customer_reorders
```

### [C5] What are the top 20 items that people put in their carts first?

```python
df_first_items = df.groupby("product_name")["add_to_cart_order"].min()
df_first_items = df_first_items.reset_index()
df_first_items.rename(columns = {"add_to_cart_order": "first_order"}, inplace = True)
df_first_items = df_first_items.sort_values(by = "first_order")
df_first_items.head(20)
```
