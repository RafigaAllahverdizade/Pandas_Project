```python
# The top 20 items that are reordered most frequently
df_reordered_products = df[df["reordered"] == 1]
df_reordered_products = df_reordered_products.groupby("product_name")["product_id"].count()
df_reordered_products = df_reordered_products.reset_index()
df_reordered_products = df_reordered_products.sort_values(by = "product_id", ascending = False)
df_reordered_products.rename(columns = {"product_id": "product_count"}, inplace = True)
df_reordered_products.head(20)
```

```python
# For each product the proportion of reorders
df_reorders = df.groupby("product_id")["reordered"].mean()
df_reorders = df_reorders.reset_index()
df_reorders.rename(columns = {"reordered": "reorder_rate"}, inplace = True)
df_reorders
```

```python
# Mean function can be used for the same purpose.
reorder_rate = df.pivot_table(index = "product_id", columns = "reordered", values = "order_id", aggfunc = "count")
reorder_rate = reorder_rate.reset_index()
reorder_rate.rename(columns = {0: "not_reordered", 1: "reordered"}, inplace = True)
reorder_rate["reorder_rate"] = reorder_rate["reordered"] / (reorder_rate["not_reordered"] + reorder_rate["reordered"])
reorder_rate = reorder_rate[["product_id", "reorder_rate"]]
reorder_rate.head()
```
